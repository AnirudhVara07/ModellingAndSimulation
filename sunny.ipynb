```python
# Latitude Flux

def calc_latitude(latitude=None, year=None):

# d1, d2 parameter vectors from 2000 to 2005

    d0 = 1.0

    d_05 = np.array([
        [ 0.109347,   -0.0129948,  -0.00831644, -0.00359449,  0.000522521, -0.00110054],  # d1
        [ 0.0101188,   0.00234080, -0.000932401, -0.00172102, -0.00156578,  0.00141373]   # d2
    ])

    # d1, d2 parameter vectors for 2006 to 2009
    d_49 = np.array([
        [ 0.144814,   0.00729394, -0.00645977, -0.00114291, -0.000587996,  0.000219460],  # d1
        [ 0.0578031, -0.0182840,   0.0123597,  -0.0122364,   0.00792947,  -0.00642885]    # d2
    ])

    if year >= 2000 and year <= 2005:
        d = d_05
    elif year >=2006 and year <= 2009:
        d = d_49
    else:
            raise ValueError("Year out of bounds")


    s = 0
    for k in range(1,7):
        
        alpha = (2*np.pi*k*latitude) / 180
 
        s += d[0,k-1] * np.cos(alpha) + d[1,k-1] * np.sin(alpha)

    Flatitude = s + d0

    return Flatitude

```

```python
def calc_longitude(longitude=None,year=None):

    g_05 = np.array([
        [-0.00477705, -0.00147749,  0.00151963,  0.000165757],  # g1
        [-0.00566262,  0.00301145,  0.0000608981, 0.0000934866]  # g2
        ])

    g_49 = np.array([
        [-0.00264432, -0.00263336,  0.00321108, -0.00180075],  # g1
        [-0.00537701, -0.00133626,  0.00121844,  0.0000279883] # g2
        ])

    s = 0
    g0 = 0

    if year >= 2000 and year <= 2005:
        g = g_05
    elif year >=2006 and year <= 2009:
        g = g_49
    else:
        raise ValueError("Year out of bounds")
    
    for k in range(1,5):
        
        alpha = (2*np.pi*k*longitude)/360

        s += g[0,k-1]*np.cos(alpha) + g[1,k-1]*np.sin(alpha)
    
    Flongitude = s + g0

    return Flongitude
```

```python

def calc_seasonal_variaton(DoY=None, year=None):
    
    b_05 = np.array([
        [ 0.209135,  -0.133610,  -0.00231834],  
        [ 0.0957844, -0.0443634,  0.0325542]    
    ])

    b_49 = np.array([
        [ 0.131082,  -0.118733,  -0.0408388],  
        [ 0.0219884, -0.0593100, -0.0137226]    
    ])

    s = 0
    b0 = 0

    if year >= 2000 and year <= 2005:
        b = b_05
    elif year >=2006 and year <= 2009:
        b = b_49
    else:
        raise ValueError("Year out of bounds")

    for k in range(1,4):

        alpha = (2*np.pi*k*DoY) / 365.25

        s = b[0,k-1]*np.cos(alpha) + b[1,k-1]*np.sin(alpha)

    Fsv = s + b0

    return Fsv

```

```python

def calc_magnetic_local(MLT=None, year=None):

    
    c_05 = np.array([
    [-0.278983,  0.0284595, -0.00449755, -0.00969936],  
    [-0.198421,  0.0430628, -0.00929224, -0.00295443]   
    ])

    c_49 = np.array([
    [-0.277790,  0.0392145, -0.000725256, 0.0152304],   
    [-0.217354,  0.0459899,  0.00473289,  0.0123554]    
    ])

    s = 0
    c0 = 0 

    if year >= 2000 and year <= 2005:
        c = c_05
    elif year >=2006 and year <= 2009:
        c = c_49
    else:
        raise ValueError("Year out of bounds")

    for k in range(1,4):

        alpha = (2*np.pi*k*MLT) / 365.25
        s = c[0,k-1]*np.cos(alpha) + c[1,k-1]*np.sin(alpha)

    Fml = s + c0

    return Fml

```
